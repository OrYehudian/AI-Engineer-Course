## Question 4: Multiply (`*`) each of the following words by the amount of its letters.
### For example: `apple` -> `'apple'` * `5 letters` -> `'appleappleappleappleapple'`
* Hi
* banana
* Encyclopedia

### Bonus: You can use the function len() to calculate the amount of letters in a word, e.g. `len('apple')` will return 5

```python
word1 = 'Hi'
word2 = 'banana'
word3 = 'Encyclopedia'
print(word1*(len(word1)))
print(word2*(len(word2)))
print(word3*(len(word3)))
```

## Question 6: Strip the following strings of spaces that are at the start or end of the string. Print the result
