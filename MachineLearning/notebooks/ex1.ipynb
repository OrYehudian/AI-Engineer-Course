```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../dataset/HousingData.csv")

df.info()
df.describe()

df.isnull().sum()

plt.figure(figsize=(6,4))
plt.boxplot(df["MEDV"])
plt.title("Boxplot of House Prices (MEDV)")
plt.ylabel("House Price")
plt.show()

df_mean = df.copy()
df_mean["LSTAT"] = df_mean["LSTAT"].fillna(df_mean["LSTAT"].mean())

df_median = df.copy()
df_median["LSTAT"] = df_median["LSTAT"].fillna(df_median["LSTAT"].median())

print(df["LSTAT"].describe())
print(df_mean["LSTAT"].describe())
print(df_median["LSTAT"].describe())

Q1 = df["CRIM"].quantile(0.25)
Q3 = df["CRIM"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["CRIM"] < lower_bound) |
    (df["CRIM"] > upper_bound)
]
print(len(outliers))
```
