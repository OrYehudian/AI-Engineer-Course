## Question 3: Given the following number `num`, calculate and print:
* The first digit
* The second digit
* The third digit
* The sum of digits

### Hint: use modulo (`%`)

```python
num = 123
firstDigit = num % 10
print(firstDigit)
secondDigit = (num // 10) % 10
print(secondDigit)
thirdDigit = (num // 100) % 10
print(thirdDigit)
sum = firstDigit + secondDigit + thirdDigit
print("The sum of digits: ", sum)
```
