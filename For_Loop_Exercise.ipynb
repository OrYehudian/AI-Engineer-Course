## Question 4:
Create a for loop that calculates and prints the factorial of `num`.

Factorial of num - The multiplication of all numbers smaller than it (including).

E.g. The factorial of 4 = $ 1 \times 2 \times 3 \times 4 = 24$

```python
num = 12
```

```python
factorial = 1
for i in range(1, num+1):
  factorial *= i
print(factorial)
```

## Question 6:
Use nested loops to print all the primal numbers from 1 to 100

Primal Number - Number that is only divided by itself and 1

```python
for num in range(2,101):
  is_prime = True

  for i in range(2,num):
    if num % i == 0:
      is_prime = False
      break
  if is_prime:
    print(num)

```

# Question 7:

Whenever you sign up or log in to a website, you're asked to enter an email address. Have you ever wondered how the site knows if the email you entered is valid?

Let’s write code to check if an email address is valid!

```python
def is_valid_email(email):
    if "@" in email and "." in email[email.index("@"):]:
        return True
    else:
        return False


email = input("Enter an email address to check if it's valid: ")

if is_valid_email(email):
    print("The email address is valid.")
else:
    print("The email address is invalid.")
```
