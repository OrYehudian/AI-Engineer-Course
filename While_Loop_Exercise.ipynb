## Question 3
Assume the user needs to guess a number between 1 and 100.

If they guess correctly – the code will print "You won!"

If they guess incorrectly: The code will print "Wrong!"

The user will be asked to guess again, and their guess will be checked again, and so on.

Can you know in advance how many attempts (guesses) the user will need to win? And what about other users? Of course not, so we need to write code that, as long as the user hasn't guessed correctly, informs them and asks them to guess again.

```python
guess = 50
x = int(input("What is your guess? (1-100) "))

while guess != x:
  print("Wrong! try again")
  x = int(input("What is your guess? (1-100) "))

print("You Won")
```

## Question 4

You want to build a shopping cart calculator for a supermarket.

The code should work for all users and all types of purchases.

Can you know in advance how many products there are or what their prices are?
Of course not, so we’ll write code that accepts prices and keeps adding them together until we tell it to stop.

When we ask the code to stop, it will print the total amount accumulated.

```python
sum = 0
itemPrice = (input("Enter ptice: "))

while itemPrice != 'stop':
  sum += float(itemPrice)
  itemPrice = (input("Enter ptice: "))

print("The sum: ", sum)
```
